```python
import pandas as pd
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
data = pd.read_csv("wine2.csv")
```

```python
print(data.head())
```

```python
x = data.iloc[:, :-1]
```

```python
x
```

```python
y = data.iloc[:,-1]
```

```python
y
```

```python
model = RandomForestClassifier()
```

```python
x.values
```

```python
model.fit(x,y)
```

```python
predictions = model.predict(x)
```

```python
predictions
```

```python
y = data.iloc[: , -1]
```

```python
actual = y.values
```

```python
count = 0
for i in range(0,len(actual)):
    if actual[i] == predictions[i]:
        count = count + 1
    print((count*100)/len(actual))
```
