```python
def shuffel_and_split_data(data,test_ratio):
    np.random.seed(42)
    shuffel_indices = np.random.permutation(len(data))
    test_set_size = int(len(data)*test_ratio)
    test_indices = shuffel_indices[:test_set_size]
    train_indicies = shuffel_indices[test_set_size]
    return data.iloc[train_indicies],data.iloc[test_indices]
```
